```python
"""
print the below pattern:
&&&&&
&&&&
&&&
&&
&
"""
n=int(input("Enter the number of rows: "))
for i in range(n, 0, -1):
    for j in range(i):
        print('&', end='')
    print()
```

```python
"""
print the below pattern:
1
22
333
4444
55555
"""
n=int(input("Enter a number between 1 to 9: "))
for i in range(1, n+1):
    for j in range(i):
        print(i, end='')
    print()
```
