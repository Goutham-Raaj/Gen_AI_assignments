```python
class Student:
    def __init__(self, name,age,course):
        self.name = name
        self.age = age
        self.course = course

    def getstudetails(self):
        print("Student name: ",self.name)
        print("Age:",self.age)
        print("Course:",self.course)
    
stu1=Student("Goutham",22,"Gen_AI")
stu2=Student("Rahul",23,"Java_Fullstack")
stu3=Student("Shiva",30,"Data_Science")
stu1.getstudetails()
stu2.getstudetails()
stu3.getstudetails()
```
