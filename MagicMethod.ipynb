```python
class Myclass:
    def __init__(self,x):
        self.x = x
    def __mul__(self,other):
        return f"Product of two numbers {self.x} and {other.x} is " +str(self.x*other.x)



obj1=Myclass(10)
obj2=Myclass(20)
print(obj1*obj2)
   
```

```python
class Myclass:
    def __init__(self,x):
        self.x = x
    def __mod__(self,other):
        return f"Remainder of two numbers {self.x} and {other.x} is " +str(self.x%other.x)



obj1=Myclass(101)
obj2=Myclass(20)
print(obj1 % obj2)
```
