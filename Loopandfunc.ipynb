```python
# print num from 10 to 1 using while
x=10
while x>0:
    print(x)
    x-=1
```

```python
#program to print multiplication table of a number
x=int(input ("Enter a number"))
for i in range(13):
    print(x,'X',i,'=',x*i)
```

```python
#program to find the sum of first n natural numbers
n=int(input("Enter a number"))
sum=0
for i in range(1,n+1):
    sum+=i
print("The sum of first", n, "natural numbers is:", sum)
```

```python
#write a function greet(name) that takes a name as input and prints a greeting message.
def greet(name):
    print("Hello, " + name)
name1 = input("Enter your name: ")
greet(name1)
```

```python
#function to add two numbers and return the result
def add_numbers(num1, num2):
    return num1 + num2
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
result = add_numbers(num1, num2)
print("The sum of", num1, "and", num2, "is:", result
```

```python
#write a function even(num) that takes a number as input and returns True if the number is even, and False otherwise.
def even(num):
    if num % 2 == 0:
        return True
    else:
        return False
num = int(input("Enter a number: "))
even_result = even(num)
print(even_result)
```

```python
#write a function to print the square of a number
def square(num):
    return num ** 2

num = int(input("Enter a number: "))
print("The square of", num, "is:", square(num))
```

```python
#write a function max_of_two(num1, num2) that takes two numbers as input and returns the maximum of the two.
def max_of_two(num1, num2):
    if num1 > num2:
        return num1
    else:
        return num2
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
max_result = max_of_two(num1, num2)
print("The maximum of", num1, "and", num2, "is:", max_result)
```
