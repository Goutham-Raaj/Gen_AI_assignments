```python
import pandas as pd
```

```python
df=pd.read_csv("C:\\Users\\Goutham\\Downloads\\Car_Sales_CSV\\car_sales.csv")
```

```python
df_new=df.copy()
```

```python
df_new["Price"] = df_new["Price"].apply(lambda x: float(x[1:].replace(",","")))
```

```python
toyota_make= df_new[(df_new["Make"] == "Toyota")]
print(toyota_make)

```

```python
budget_car= df_new[(df_new["Price"]<=5000.0)]
print(budget_car)
```
