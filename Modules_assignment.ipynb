```python
#write a python program to find the length of a  list using len()
a=[1, 2, 3, 4, 5]
length = len(a)
print("The length of the list is:", length)
```

```python
#Write a python program to find the maximum and minimum value in a list using max() and min()
a=[1, 2, 3, 4, 5]
maximum = max(a)
minimum = min(a)
print("The maximum value in the list is:", maximum)
print("The minimum value in the list is:", minimum)
```

```python
#use sum() to find the sum of all the elements in a list
a=[1, 2, 3, 4, 5]
total = sum(a)
print("The sum of all the elements in the list is:", total)
```

```python
#create a list and sort it in ascending order and descending order using sort()
a=[5, 2, 9, 1, 5, 6]
print("Original list:", a)

a.sort()
print("List sorted in ascending order:", a)

a.sort(reverse=True)
print("List sorted in descending order:", a)
```

```python
#write a python program to create a tuple and print it 
a=(1, 2, 3, 4, 5)
print("The tuple is:", a)
```

```python
#write a python program to access the first, middle and last element of a tuple using indexing
a=(1, 2, 3, 4, 5, 6)
print("The first element of the tuple is:", a[0])
print("The middle element of the tuple is:", a[len(a)//2])
print("The last element of the tuple is:", a[-1])
```

```python
#write a python program to remove an element fromm a set using discard()
a={1, 2, 3, 4, 5}
a.discard(3)
print("The set after removing an element is:", a)
```

```python
#write a python program to access the first three and last three elements of a tuple using slicing
a=(1, 2, 3, 4, 5, 6)
print("The first three elements of the tuple are:", a[:3])
print("The last three elements of the tuple are:", a[-3:])
```

```python
#write a python program to remove an random element from a set using pop()
a={1, 2, 3, 4, 5}
a.pop()
print("The set after removing a random element is:", a)
```

```python
#write a python program to update the value of a existing key in a dictionary
a={'a': 1, 'b': 2, 'c': 3}
a['b'] = 5
print("The dictionary after updating the value of key 'b' is:", a)
```

```python
#write a python program to remove the last inserted key-value pair from a dictionary using popitem()
a={'a': 1, 'b': 2, 'c': 3}
a.popitem()
print("The dictionary after removing the last inserted key-value pair is:", a)
```

```python
#write a python program to remove all key-value pairs from a dictionary using clear()
a={'a': 1, 'b': 2, 'c': 3}
a.clear()
print("The dictionary after removing all key-value pairs is:", a)
```

```python
#write a python program to get all keys from a dictionary using keys()
a={'a': 1, 'b': 2, 'c': 3}
print("The keys in the dictionary are:", list(a.keys()))
```

```python
#write a python program to list all key_value pairs in a dictionary using items()
a={'a': 1, 'b': 2, 'c': 3}
print("The key-value pairs in the dictionary are:", list(a.items()))
```

```python
#write a python program to check if a key is present in a dictionary using in
a={'a': 1, 'b': 2, 'c': 3}
n=str(input("Enter the key to check: "))
if n in a:
    print(f"Key '{n}' is present in the dictionary.")
else:
    print(f"Key '{n}' is not present in the dictionary.")
```
